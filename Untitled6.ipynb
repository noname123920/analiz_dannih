```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('credit_risk_dataset.csv')
df.head()
```

```python
table1 = df.pivot_table(
    values='loan_amnt',
    index='person_home_ownership',
    columns='loan_intent',
    aggfunc='median'
).rename_axis(index=None, columns=None)

table1
```

```python
table2 = df.pivot_table(
    values='person_income',
    index='person_home_ownership',
    columns='loan_intent',
    aggfunc='median'
).rename_axis(index=None, columns=None)

table2
```

```python
sns.heatmap(table1, annot=True, fmt='.0f', cmap='YlOrRd')
plt.show()
```

```python
sns.heatmap(table2, annot=True, fmt='.0f', cmap='YlOrRd')
plt.show()
```
