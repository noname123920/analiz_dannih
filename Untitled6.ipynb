```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('credit_risk_dataset.csv')
df.head()
```

```python
df[['person_home_ownership', 'loan_intent', 'loan_amnt', 'person_income']].head()
```

```python
pivot1 = df.pivot_table(values='loan_amnt', index='loan_intent', columns='person_home_ownership', aggfunc='median')
pivot1
```

```python
pivot2 = df.pivot_table(values='person_income', index='loan_intent', columns='person_home_ownership', aggfunc='median')
pivot2
```

```python
sns.heatmap(pivot1)
plt.show()
```

```python
sns.heatmap(pivot2)
plt.show()
```
