```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.model_selection import train_test_split
```

```python
df2 = pd.read_csv('student-mat.csv', sep=';')
df2.head()
```

```python
df2_num = df2.select_dtypes(include="number")
df2_num.head()
```

```python
corr_matrix = df2_num.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Матрица корреляций числовых признаков')
plt.tight_layout()
plt.show()
```

```python
df2_num['y'] = (df2_num['G3'] >= 12).astype(int)
df2_num[['G3', 'y']].head()
```

## Без G1 и G2:

```python
X = df2_num.drop(columns='G3')
y = df2_num['y']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=42)
```

```python
X_train_v1 = X_train.drop(columns=['G1', 'G2'])
X_test_v1 = X_test.drop(columns=['G1', 'G2'])

model_v1 = LogisticRegression(max_iter=106).fit(X_train_v1, y_train)
```

```python
y_pred_v1 = model_v1.predict(X_test_v1)
display(f'{accuracy_score(y_test, y_pred_v1) = }')
```

```python
ConfusionMatrixDisplay.from_predictions(y_true=y_test, y_pred=y_pred_v1, cmap='hot')
plt.title('Матрица ошибок (Вариант 1)')
plt.show()
```

## Вместе с G1 и G2:

```python
model_v2 = LogisticRegression(max_iter=106).fit(X_train, y_train)
```

```python
y_pred_v2 = model_v2.predict(X_test)
display(f'{accuracy_score(y_test, y_pred_v1) = }')
```

```python
ConfusionMatrixDisplay.from_predictions(y_true=y_test, y_pred=y_pred_v2, cmap='hot')
plt.title('Матрица ошибок (Вариант 2)')
plt.show()
```
