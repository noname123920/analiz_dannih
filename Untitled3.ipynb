```python
import numpy as np
import os
from scipy import stats
import matplotlib.pyplot as plt
```

```python
print(os.listdir())
```

```python
data = np.loadtxt('cholestr.txt', delimiter='\t', skiprows=1, usecols=0)
data.size
```

```python
# help(stats.ttest_1samp)
```

```python
statistic, pvalue = stats.ttest_1samp(data, popmean=240)
statistic, pvalue
```

```python
alpha = 0.05 # уровень значимости
if pvalue < alpha:
    print("Гипотеза плохо согласуется с наблюдаемыми данными")
else:
    print("Нет оснований отвергнуть гипотезу при уровне значимости 0.05")
```

```python
t = float(statistic)
t
```

```python
df = data.size - 1
```

```python
x = np.linspace(-5, 5, 500)
```

```python
y = stats.t.pdf(x, df)
```

```python
plt.plot(x, y)

x_right = x[x >= t]
y_right = y[x >= t]
plt.fill_between(x_right, y_right, color='red', alpha=0.4)

x_left = x[x <= -t]
y_left = y[x <= -t]
plt.fill_between(x_left, y_left, color='red', alpha=0.4)

plt.axvline(t, color='red')
plt.axvline(-t, color='red')
plt.show()
```
