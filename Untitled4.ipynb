```python
x = df['HEIGHT'].values
nx = len(x)
nx
```

```python
k = 15
param = 2

c = 1
m, sigma = x.mean(), x.std(ddof=1)
a, b = m - c*sigma, m + c*sigma
hist = np.histogram(x, bins=k-2, range=(a, b))
print(f'{hist = }')

f_obs = hist[0]

f_obs = np.hstack([
    np.count_nonzero(x < a), f_obs, np.count_nonzero(x > b)
])
print(f'{f_obs = }')
print(f'{f_obs.sum() = :d}')
```

```python
np.mean((x >= a) & (x <= b)) * 100
```

```python
np.count_nonzero(x < a)
```

```python
np.count_nonzero(x > b)
```

```python
f_exp = np.zeros(k)
edges = hist[1]
print(f'{edges = }')
N = stats.norm(loc=m, scale=sigma)
f_exp[0] = N.cdf(edges[0])
for j in range(1, k-1):
  f_exp[j] = N.cdf(edges[j]) - N.cdf(edges[j-1])
f_exp[-1] = N.sf(edges[-1])
f_exp *= nx
print(f'{f_exp = }')
print(f'{f_exp.sum() = :f}')
```

```python
stats.chisquare(f_obs, f_exp, ddof=param)
```

```python
df_delete = df.drop([38, 41])
print(df_delete.shape)
```

```python
plt.hist(df_delete['HEIGHT'])
plt.xlabel('Рост')
plt.ylabel('Частота')
plt.title('Гистограмма роста')
plt.show()
```

```python
x_delete = df_delete['HEIGHT'].values
nx_delete = len(x_delete)
nx_delete
```

```python
stats.normaltest(x_delete)
```

```python
k_new = 12
param = 2

c = 1
m_c, sigma_c = x_delete.mean(), x_delete.std(ddof=1)
a_c, b_c = m_c - c*sigma_c, m_c + c*sigma_c
hist_c = np.histogram(x_delete, bins=k_new-2, range=(a_c, b_c))
print(f'{hist_c = }')

f_obs_c = hist_c[0]

f_obs_c = np.hstack([
    np.count_nonzero(x_delete < a_c), f_obs_c, np.count_nonzero(x_delete > b_c)
])

print(f'{f_obs_c = }')
print(f'{f_obs_c.sum() = :d}')
```

```python
f_exp_c = np.zeros(k_new)
edges_c = hist_c[1]
print(f'{edges_c = }')
N_c = stats.norm(loc=m_c, scale=sigma_c)
f_exp_c[0] = N_c.cdf(edges_c[0])
for j in range(1, k_new-1):
    f_exp_c[j] = N_c.cdf(edges_c[j]) - N_c.cdf(edges_c[j-1])
f_exp_c[-1] = N_c.sf(edges_c[-1])
f_exp_c *= nx_delete
print(f'{f_exp_c = }')
print(f'{f_exp_c.sum() = :f}')
```

```python
stats.chisquare(f_obs_c, f_exp_c, ddof=param)
```
