```python
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
print(f"Классы: {set(y)}")
```

## One-vs-Rest

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.metrics import accuracy_score
```

```python
ovr = OneVsRestClassifier(LogisticRegression())
ovr.fit(X_train, y_train)
y_pred_ovr = ovr.predict(X_test)
```

```python
print("OvR accuracy:", accuracy_score(y_test, y_pred_ovr))
print("Число обученных моделей:", len(ovr.estimators_))
```

## One-Vs-One

```python
from sklearn.multiclass import OneVsOneClassifier
```

```python
ovo = OneVsOneClassifier(LogisticRegression())
ovo.fit(X_train, y_train)
y_pred_ovo = ovo.predict(X_test)
```

```python
print("OvO accuracy:", accuracy_score(y_test, y_pred_ovo))
print("Число обученных моделей:", len(ovo.estimators_)) # k * (k-1) / 2. k - количество классов
```
