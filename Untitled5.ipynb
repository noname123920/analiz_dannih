```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
eyes = pd.read_csv('eysenck.txt', sep='\t')
eyes.head()
```

```python
young = eyes[eyes['Age'] == 'Younger']
young.size
```

```python
sns.boxplot(x='Process', y='Words', data=young)
plt.xlabel('Process')
plt.ylabel('Words')
plt.show()
```

```python
young.info()
```

```python
y = young.groupby('Process')['Words'].apply(list)
y
```

```python
stats.f_oneway(*y)
```

```python
stats.alexandergovern(*y)
```

```python
stats.kruskal(*y)
```

```python
old = eyes[eyes['Age'] == 'Older']
old.head()
```

```python
sns.boxplot(x='Process', y='Words', data=old)
plt.xlabel('Process')
plt.ylabel('Words')
plt.show()
```

```python
o = old.groupby('Process')['Words'].apply(list)
o
```

```python
stats.f_oneway(*o)
```

```python
stats.alexandergovern(*o)
```

```python
stats.kruskal(*o)
```
