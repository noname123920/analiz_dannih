```python
import numpy as np
from scipy import stats
import os
```

```python
print(os.listdir())
```

```python
data = np.loadtxt('abalone.data', delimiter=',', dtype=str)
data
```

```python
population = data[:, 1].astype(float)
population
```

```python
population.size
```

```python
pop_mean = np.mean(population)
pop_mean
```

```python
pop_median = np.median(population)
pop_median
```

```python
rng = np.random.default_rng(seed=42)
rng
```

```python
r = 10
```

```python
sample = rng.choice(population, size=r, replace=False)
sample
```

```python
sample_mean = np.mean(sample)
sample_mean
```

```python
sample_std = np.std(sample, ddof=1)
sample_std
```

```python
se = stats.sem(sample)
se
```

```python
ci_t = stats.t.interval(0.90, df=r-1, loc=sample_mean, scale=se)
ci_t
```

```python
t = (ci_t[1] - sample_mean) / se
print(t)
```

```python
ci_t_right = sample_mean + t * se
ci_t_right
```

```python
ci_t_left = sample_mean - t * se
ci_t_left
```

```python
stats.t.ppf((1 - (1 - 0.9) / 2), df=r-1)
```

```python
N = 10000
```

```python
count_t = 0
```

```python
for i in range(N):
  sample = rng.choice(population, size=r, replace=True)
  sample_mean = np.mean(sample)
  se = stats.sem(sample)
  ci_t = stats.t.interval(0.90, df=r-1, loc=sample_mean, scale=se)
  if ci_t[0] <= pop_mean <= ci_t[1]:
    count_t += 1
```

```python
percent_t = count_t / N
percent_t
```

```python
count_norm = 0
```

```python
for i in range(N):
  sample = rng.choice(population, size=r, replace=True)
  sample_mean = np.mean(sample)
  se = stats.sem(sample)
  ci_norm = stats.norm.interval(0.90, loc=sample_mean, scale=se)
  if ci_norm[0] <= pop_mean <= ci_norm[1]:
    count_norm += 1
```

```python
percent_norm = count_norm / N
percent_norm
```

```python
ci_norm = stats.norm.interval(0.90, loc=sample_mean, scale=se)
ci_norm
```

```python
z = (ci_norm[1] - sample_mean) / se
print(z)
```

```python
stats.norm.ppf((1 - (1 - 0.9) / 2))
```

```python
n = 100
N = 500
r = 50
```

```python
sample = rng.choice(population, size=n, replace=False)
sample.size
```

```python
count_boot = 0
```

```python
for i in range(10000):
  sample = rng.choice(population, size=n, replace=True)
  ci_boot = stats.bootstrap((sample,),  statistic=np.median, method='percentile', confidence_level=0.9, n_resamples=N)
  res = ci_boot.confidence_interval
  if res.low <= pop_median <= res.high:
    count_boot += 1
```

```python
percent_boot = count_boot / 10000
percent_boot
```

```python
print(ci_boot)
```

```python
res = ci_boot.confidence_interval
res
```
