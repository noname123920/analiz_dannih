```python
df = pd.read_csv('default_of_credit_card_clients.csv', skiprows=1, sep='\t')
df.columns = ['ID'] + [f'X{i}' for i in range(1, 24)] + ['Y']
df
```

```python
credit_payers = df.loc[df['Y'] == 0, 'X1']
credit_payers.shape
```

```python
credit_default = df.loc[df['Y'] == 1, 'X1']
credit_default.shape
```

```python
plt.hist(credit_payers, density=True, color='blue', label='Платящие (Y=0)')
plt.hist(credit_default, density=True, color='red', label='Неплатящие (Y=1)', alpha=0.5)
plt.ticklabel_format(style='plain', axis='both')
plt.xlabel('Кредит X1 (тайваньские доллары)')
plt.ylabel('Плотность вероятности')
plt.legend()
plt.show()
```

```python
print("Средний кредит платящих: ", credit_payers.mean())
print("Средний кредит неплатящих: ", credit_default.mean())
```
