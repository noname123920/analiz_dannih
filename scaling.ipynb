```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from scipy.stats import iqr
```

```python
df = pd.read_csv('example_2.csv')
display(df)
```

$$\hat{x}=\frac{x-\overline{x}}{\sigma_x}$$

```python
scaled = StandardScaler().fit_transform(df)
scaled
```

```python
scaled.mean(axis=0), scaled.std(axis=0)
```

$$\hat{x}=\frac{x-x_{min}}{x_{max}-x_{min}}$$

```python
MinMaxScaler().fit_transform(df)
```

$$\hat{x}=\frac{x-Me(\textbf{x})}{IQR(\textbf{x})}$$

```python
scaled = RobustScaler().fit_transform(df)
scaled
```

```python
np.median(scaled, axis=0), iqr(scaled, axis=0)
```
