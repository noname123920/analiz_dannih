```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer, KNNImputer
```

```python
df = pd.read_csv('example.csv')
display(df)
df.info()
```

```python
df.dropna(axis=0)
```

```python
df.dropna(subset=['C'], how='any')
```

```python
imputer = SimpleImputer().fit(df)
imputer.transform(df)
```

```python
df2 = pd.DataFrame(data=[[np.nan]*4], columns=df.columns)
df2
```

```python
imputer.transform(df2)
```

```python
SimpleImputer(missing_values=np.nan, strategy='constant', fill_value=-999).fit_transform(df)
```

```python
SimpleImputer(missing_values=np.nan, strategy='median').fit_transform(df)
```

```python
SimpleImputer(missing_values=np.nan, strategy='most_frequent').fit_transform(df)
```

```python
df
```

```python
KNNImputer(n_neighbors=2, metric='nan_euclidean').fit_transform(df)
```
